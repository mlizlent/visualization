```python
import numpy as np
import pandas as pd
import plotly.graph_objects as go
from plotly.offline import init_notebook_mode

# Enable offline rendering in classic Jupyter Notebooks if needed
init_notebook_mode(connected=True)

# Generate data
returns = np.random.randn(50)
price = 100 + np.cumsum(returns)
dates = pd.date_range(start='2020-09-01', periods=50, freq='B')

df = pd.DataFrame(zip(dates, price), columns=['Date', 'Company_A'])

# Create figure with a Scatter line trace
fig = go.Figure(
    data=go.Scatter(
        x=df['Date'],
        y=df['Company_A'],
        mode='lines',
        name='Company_A'
    )
)

# Add title and axis labels
fig.update_layout(
    title='Company_1 Stock Price',
    xaxis_title='Date',
    yaxis_title='Company_A'
)

# Display the plot
fig.show()
```
